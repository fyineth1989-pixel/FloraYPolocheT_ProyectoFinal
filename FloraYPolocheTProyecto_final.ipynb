6. Seleccione dos variables que podrían estar relacionadas dentro del contexto del problema. Convierta las series correspondientes a estas variables en vectores de NumPy e implemente una regresión lineal utilizando numpy.polyfit().

Paso 1: Seleccionar dos variables relacionadas

Una relación natural para explorar sería:

Mes_contagio (numérico, de 1 a 12)

Número de contagios en cada mes

Esto nos permitirá ver si existe una tendencia lineal en los contagios a lo largo del año.

```python
# Paso 2: Convertir a vectores NumPy

# Usar el conteo mensual ya calculado
conteo_mes = df["Mes_contagio"].value_counts().reindex([
    "ENERO","FEBRERO","MARZO","ABRIL","MAYO","JUNIO",
    "JULIO","AGOSTO","SEPTIEMBRE","OCTUBRE","NOVIEMBRE","DICIEMBRE"
]).fillna(0)

# Crear vectores NumPy
x = np.arange(1, 13)                # Meses del 1 al 12
y = conteo_mes.values.astype(float) # Número de contagios
```

```python
# Paso 3: Regresión lineal con numpy.polyfit()

# Ajustar regresión lineal (grado 1)
coef = np.polyfit(x, y, 1)  # [pendiente, intercepto]
pendiente, intercepto = coef

# Crear recta estimada
y_pred = np.polyval(coef, x)

print("Pendiente:", pendiente)
print("Intercepto:", intercepto)
```

```python
# Paso 4: Visualización

plt.figure(figsize=(10,6))
plt.scatter(x, y, color="blue", label="Datos reales")
plt.plot(x, y_pred, color="red", linewidth=2, label="Tendencia lineal")

plt.title("Regresión lineal de contagios por mes", fontsize=14, fontweight="bold")
plt.xlabel("Mes (1=Enero ... 12=Diciembre)")
plt.ylabel("Número de contagios")
plt.xticks(x, conteo_mes.index, rotation=45)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

Contraste de hipótesis

Hipótesis nula (H₀): La edad promedio de contagio es igual en hombres y mujeres.

Hipótesis alternativa (H₁):

```python
import numpy as np
from scipy import stats
import pandas as pd

# carga de base de datos
df = pd.read_csv(r"C:\Users\fyine\Documents\U.Compensar\Programacion para ciencias de datos ll\Eventos_Epidemiológicos_20250822.CSV",
                 sep=None, engine="python", encoding="utf-8-sig")

# Verificar las primeras filas
print(df.head())
```

# Regresión lineal

1. Utilice la función train_test_split de la biblioteca scikit-learn para dividir sus datos en un conjunto de entrenamiento y un conjunto de prueba. Es esencial que al menos el 70% de los datos se utilicen para entrenar el modelo. Esta proporción, conocida como la división 70-30, proporciona un equilibrio entre la cantidad de datos disponibles para entrenar y para evaluar el modelo. Esta elección debe estar justificada y puede ajustarse según las características específicas de su conjunto de datos.

# Respuesta

```python
# Orden de meses
orden = ["ENERO","FEBRERO","MARZO","ABRIL","MAYO","JUNIO",
         "JULIO","AGOSTO","SEPTIEMBRE","OCTUBRE","NOVIEMBRE","DICIEMBRE"]
map_mes = {m:i+1 for i,m in enumerate(orden)}

# Normalizar mes y crear variable numérica
tmp = df.copy()
tmp["Mes_contagio"] = tmp["Mes_contagio"].astype(str).str.upper().str.strip()
tmp = tmp[tmp["Mes_contagio"].isin(orden)].copy()
tmp["mes_num"] = tmp["Mes_contagio"].map(map_mes)

# Agregar variables por mes
agg = tmp.groupby("Mes_contagio").agg(
    n_contagios=("Edad_paciente","count"),
    edad_media=("Edad_paciente","mean"),
    prop_fem=("Sexo_paciente", lambda s: (s=="F").mean()),
    mes_num=("mes_num","first")
).reindex(orden)

display(agg)
```

```python
from sklearn.model_selection import train_test_split

# Definir variables
X = agg[["mes_num","edad_media","prop_fem"]].fillna(0)
y = agg["n_contagios"].values

# División 70–30
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

print("Tamaño entrenamiento:", X_train.shape[0])
print("Tamaño prueba:", X_test.shape[0])
```

Justificación de la proporción 70–30

La proporción 70–30 es muy popular porque logra encontrar un buen balance entre el conjunto de entrenamiento, que es lo suficientemente grande para que el modelo aprenda bien, y el conjunto de prueba, que debe ser representativo para evaluar qué tan bien funciona. Al usar esta proporción, se reduce el riesgo de que el modelo se ajuste demasiado a los datos de entrenamiento y también se obtiene una evaluación más justa y realista del rendimiento del modelo.

1. Divida los datos en un conjunto de entrenamiento y un conjunto de prueba utilizando la función train_test_split de scikit-learn. Asegúrese de que al menos el 70% de los datos se utilicen para entrenar el modelo, como es común en problemas de clasificación.

# Respuesta

```python
from sklearn.model_selection import train_test_split

# Variable independiente (X): edad
X = df[["Edad_paciente"]].fillna(0)

# Variable dependiente (y): sexo (1 = mujer, 0 = hombre)
y = (df["Sexo_paciente"] == "F").astype(int)

# Dividir los datos en entrenamiento (70%) y prueba (30%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

print("Tamaño del conjunto de entrenamiento:", X_train.shape[0])
print("Tamaño del conjunto de prueba:", X_test.shape[0])
```

2. Utilice la clase LogisticRegression de scikit-learn para crear un modelo de regresión logística. Recuerde que la regresión logística es ideal para problemas de clasificación binaria.

# Respuesta

```python
from sklearn.linear_model import LogisticRegression

# 1) Crear el modelo logístico
log_model = LogisticRegression()

# 2) Entrenar el modelo con los datos de entrenamiento
log_model.fit(X_train, y_train)

print("Modelo de regresión logística creado y entrenado con éxito")
```

3. Ajuste el modelo a los datos de entrenamiento utilizando el método fit. Prueba con diferentes valores de C (parametro de regularización) y diferentes algoritmos de optimización (solver). Redacta un párrafo en el que expliques como afectan estas opciones al desempeño de la optimización.  

# Respuesta

```python
from sklearn.linear_model import LogisticRegression

# Probar con distintos valores de C y solver
for C in [0.1, 1.0, 10]:
    for solver in ["liblinear", "lbfgs"]:
        log_model = LogisticRegression(C=C, solver=solver, max_iter=1000)
        log_model.fit(X_train, y_train)
        score = log_model.score(X_test, y_test)
        print(f"C={C}, solver={solver} → Precisión en prueba: {score:.3f}")
```

Al ajustar el modelo usando diferentes valores de C y distintos algoritmos de optimización (solver), podemos ver que estas opciones influyen en cómo funciona el modelo. Cuando eliges valores pequeños de C, estás aplicando más regularización, lo que ayuda a simplificar el modelo y puede evitar que se sobreajuste a los datos, aunque esto también podría hacer que la precisión disminuya un poco. Por otro lado, si usas valores grandes de C, la regularización se reduce, permitiendo que el modelo se ajuste mejor a los datos de entrenamiento, pero con el riesgo de que se adapte demasiado y no generalice bien. Sobre el solver, algoritmos como liblinear funcionan bien con conjuntos de datos pequeños o con variables binarias, mientras que lbfgs es más eficiente para conjuntos de datos grandes y problemas con varias clases. En definitiva, elegir los parámetros adecuados es buscar un balance entre obtener una buena precisión y que el modelo pueda aplicar ese conocimiento a nuevos datos de forma efectiva.

8. Calcula la precisión del modelo en el conjunto de prueba. La precisión es importante, pero también considera otras métricas como la sensibilidad y la especificidad para obtener una imagen completa del rendimiento del modelo.   

# Respuesta

```python
from sklearn.metrics import accuracy_score, recall_score, confusion_matrix

# Precisión (accuracy): aciertos totales / total
accuracy = accuracy_score(y_test, y_pred)

# Sensibilidad (recall para la clase positiva = 1)
sensibilidad = recall_score(y_test, y_pred)

# Especificidad = verdaderos negativos / (verdaderos negativos + falsos positivos)
cm = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cm.ravel()
especificidad = tn / (tn + fp)

print("Precisión (Accuracy):", accuracy)
print("Sensibilidad (Recall):", sensibilidad)
print("Especificidad:", especificidad)
```

El modelo logró alcanzar una precisión del X%, lo que significa que esa proporción de las predicciones del modelo fueron correctas en el conjunto de prueba. La sensibilidad nos indica qué tan bien el modelo pudo detectar correctamente los casos positivos, mientras que la especificidad muestra qué tan efectivo fue para reconocer los casos negativos. Estas métricas trabajan juntas con la precisión para darnos una idea más completa del rendimiento del modelo, ya que a veces uno puede ser alto mientras que el otro puede fallar, especialmente en la detección de diferentes clases de datos.

```python

!pip install dash plotly gdown --quiet

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from dash import Dash, dcc, html, Input, Output, ctx
import gdown
import dash

file_id = "14rePLTS1h3yS5qRedrqSz7HVAH_TX4ND"
url = f"https://drive.google.com/uc?id={file_id}"
csv_path = "Eventos_Epidemiológicos_20250822.CSV"
gdown.download(url, csv_path, quiet=False)

df = pd.read_csv(csv_path, encoding="utf-8-sig")


orden_meses = [
    "ENERO","FEBRERO","MARZO","ABRIL","MAYO","JUNIO",
    "JULIO","AGOSTO","SEPTIEMBRE","OCTUBRE","NOVIEMBRE","DICIEMBRE"
]
df["Mes_contagio"] = df["Mes_contagio"].str.upper()
df["Mes_contagio"] = pd.Categorical(df["Mes_contagio"], categories=orden_meses, ordered=True)


def kpi_card(title, value):
    return html.Div(
        style={
            "background": "#f8fafc",
            "borderRadius": "10px",
            "padding": "10px",
            "textAlign": "center",
            "boxShadow": "0 2px 6px rgba(0,0,0,0.1)"
        },
        children=[
            html.Div(title, style={"fontSize":"13px","color":"#64748b"}),
            html.Div(value, style={"fontSize":"20px","fontWeight":"bold","color":"#0f172a"})
        ]
    )


app = Dash(__name__)

app.layout = html.Div([
    html.H2("Dashboard Epidemiológico - Interactivo", style={"textAlign": "center"}),

    # Narrativa introductoria
    html.P("Este tablero interactivo permite explorar la evolución de los eventos epidemiológicos "
           "durante el año, filtrando por sexo y mes para identificar patrones, tendencias y posibles "
           "factores de riesgo. Además, incluye análisis de regresión para observar comportamientos generales.",
           style={"textAlign": "center", "fontSize": "14px", "color": "#475569", "maxWidth": "800px", "margin": "auto"}),

    # Filtros
    html.Div([
        html.Label("Sexo"),
        dcc.Dropdown(
            id="filtro-sexo",
            options=[{"label": s, "value": s} for s in sorted(df["Sexo_paciente"].dropna().unique())],
            value=None, placeholder="Todos"
        ),
        html.Label("Mes(es)"),
        dcc.Dropdown(
            id="filtro-mes",
            options=[{"label": m, "value": m} for m in orden_meses],
            value=[], multi=True, placeholder="Todos"
        )
    ], style={"display": "flex", "gap": "20px", "marginBottom": "20px"}),

    # Opciones de escala y tipo de gráfico
    html.Div([
        html.Label("Escala del eje Y:"),
        dcc.RadioItems(
            id="escala-y",
            options=[
                {"label": "Lineal", "value": "linear"},
                {"label": "Logarítmica", "value": "log"}
            ],
            value="linear",
            inline=True
        ),
        html.Label("Tipo de gráfico:"),
        dcc.Dropdown(
            id="tipo-grafico",
            options=[
                {"label": "Línea", "value": "line"},
                {"label": "Barras", "value": "bar"},
                {"label": "Área", "value": "area"}
            ],
            value="line",
            clearable=False,
            style={"width": "200px"}
        )
    ], style={"display": "flex", "gap": "20px", "marginBottom": "20px"}),

    # KPIs
    html.Div(style={"display":"grid","gridTemplateColumns":"repeat(3, 1fr)","gap":"10px","marginBottom":"16px"}, children=[
        html.Div(id="kpi-total"),
        html.Div(id="kpi-edad"),
        html.Div(id="kpi-mesmax")
    ]),

    # Narrativa dinámica
    html.Div(id="texto-narrativa", style={"marginTop": "10px", "textAlign": "center", "fontSize": "14px", "color": "#334155"}),

    # Gráficas lado a lado
    html.Div(style={"display": "grid", "gridTemplateColumns": "1fr 1fr", "gap": "10px"}, children=[
        dcc.Graph(id="grafico-linea", style={"height":"420px"}),
        dcc.Graph(id="grafico-regresion", style={"height":"420px"})
    ]),

    # Conclusiones finales
    html.H4("Conclusiones generales", style={"marginTop": "30px"}),
    html.P("El análisis muestra picos en ciertos meses que podrían relacionarse con factores ambientales "
           "o eventos específicos. La línea de regresión indica una tendencia general a la disminución "
           "de casos hacia fin de año, aunque con variaciones mensuales significativas.",
           style={"fontSize": "14px", "color": "#475569", "maxWidth": "900px", "margin": "auto", "textAlign": "justify"}),

    # Botón de descarga
    html.Button("Descargar CSV", id="btn-descargar"),
    dcc.Download(id="download-datos")
])


@app.callback(
    Output("kpi-total", "children"),
    Output("kpi-edad", "children"),
    Output("kpi-mesmax", "children"),
    Output("grafico-linea", "figure"),
    Output("grafico-regresion", "figure"),
    Output("download-datos", "data"),
    Output("texto-narrativa", "children"),
    Input("filtro-sexo", "value"),
    Input("filtro-mes", "value"),
    Input("escala-y", "value"),
    Input("tipo-grafico", "value"),
    Input("btn-descargar", "n_clicks")
)
def actualizar(sexo, meses, escala_y, tipo_grafico, n_clicks):
    d = df.copy()
    if sexo:
        d = d[d["Sexo_paciente"] == sexo]
    if meses:
        d = d[d["Mes_contagio"].isin(meses)]

    # KPIs dinámicos
    total_casos = len(d)
    edad_prom = d["Edad_paciente"].mean() if not d.empty else 0
    mes_max = d["Mes_contagio"].value_counts().idxmax() if not d.empty else "-"

    kpi1 = kpi_card("Total de casos", f"{total_casos:,}".replace(",", "."))
    kpi2 = kpi_card("Edad promedio", f"{edad_prom:.2f}")
    kpi3 = kpi_card("Mes con más casos", mes_max)

    # Narrativa dinámica
    if total_casos == 0:
        narrativa = "No hay datos disponibles para los filtros seleccionados."
    else:
        narrativa = f"Se registraron {total_casos} casos en el período seleccionado. El mes con mayor número de contagios fue {mes_max}."

    # Datos para gráficas
    conteo_mes = d["Mes_contagio"].value_counts().reindex(orden_meses, fill_value=0).astype(float)
    x = conteo_mes.index
    y = conteo_mes.values

    # Gráfico principal dinámico
    fig_line = go.Figure()
    if tipo_grafico == "line":
        fig_line.add_trace(go.Scatter(x=x, y=y, mode="lines+markers", fill=None))
    elif tipo_grafico == "bar":
        fig_line.add_trace(go.Bar(x=x, y=y))
    elif tipo_grafico == "area":
        fig_line.add_trace(go.Scatter(x=x, y=y, mode="lines", fill="tozeroy"))

    fig_line.update_layout(
        yaxis_type=escala_y,
        yaxis=dict(rangemode="nonnegative"),
        height=420
    )

    # Gráfico de regresión lineal
    if len(y) >= 2 and np.any(y):
        coef = np.polyfit(np.arange(1, len(y)+1), y, 1)
        y_pred = np.poly1d(coef)(np.arange(1, len(y)+1))
    else:
        y_pred = np.zeros_like(y)

    fig_reg = go.Figure()
    fig_reg.add_trace(go.Scatter(x=x, y=y, mode="markers", name="Datos reales"))
    fig_reg.add_trace(go.Scatter(x=x, y=y_pred, mode="lines", name="Regresión lineal", line=dict(color="red")))
    fig_reg.update_layout(
        yaxis_type=escala_y,
        yaxis=dict(rangemode="nonnegative"),
        height=420
    )

    # Exportar CSV al hacer clic
    data_download = None
    if ctx.triggered_id == "btn-descargar":
        data_download = d.to_csv(index=False)

    return kpi1, kpi2, kpi3, fig_line, fig_reg, (dash.dcc.send_string(data_download, "datos_filtrados.csv") if data_download else None), narrativa



from google.colab import output as colab_output
PORT = 8050
colab_output.serve_kernel_port_as_window(PORT, path='/')
app.run(host='0.0.0.0', port=PORT, debug=False)
```
